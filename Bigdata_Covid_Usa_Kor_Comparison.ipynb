```python
list(raw_df)
# location date total_cases (need)
select_col1 = ['location','date','total_cases']
revised_df = raw_df[select_col1]
#원래 있던 행에서 가져오고 싶은것들
revised_df.shape
target_df = revised_df[(revised_df['location']== "South Korea")| (revised_df['location']=="United States")]
target_df
```

## Comparison of USA  KOR


```python
target_df['date'] = pd.to_datetime(target_df['date'])
plt.figure(figsize=(10,5))
sns.lineplot(data=target_df,x='date',y='total_cases',hue='location',linewidth='2')
plt.title('covid trend')
plt.xlabel('date')
plt.ylabel('num')
plt.grid(True)
plt.xticks(rotation=45)#회전변환을왜할까
plt.show()#그래프 표현
```
